```python

import sqlite3
import pandas as pd
from datetime import datetime

# =========================
# 1. 建立資料庫
# =========================

conn = sqlite3.connect("/content/stock_manager.db")
conn.execute("PRAGMA foreign_keys = ON")

conn.executescript("""
CREATE TABLE IF NOT EXISTS stocks (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    market TEXT NOT NULL,
    symbol TEXT NOT NULL,
    name TEXT NOT NULL,
    sector TEXT DEFAULT '',
    created_at TEXT NOT NULL,
    UNIQUE(market, symbol)
);

CREATE TABLE IF NOT EXISTS trades (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    stock_id INTEGER NOT NULL,
    trade_date TEXT NOT NULL,
    side TEXT NOT NULL,
    quantity REAL NOT NULL,
    price REAL NOT NULL,
    fee REAL DEFAULT 0,
    note TEXT DEFAULT '',
    FOREIGN KEY(stock_id) REFERENCES stocks(id)
);

CREATE TABLE IF NOT EXISTS notes (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    stock_id INTEGER,
    note_date TEXT NOT NULL,
    title TEXT NOT NULL,
    content TEXT NOT NULL,
    FOREIGN KEY(stock_id) REFERENCES stocks(id)
);
""")

conn.commit()
print("股票管理系統資料庫已建立！")


# =========================
# 2. 新增股票
# =========================

def add_stock(market, symbol, name, sector=""):
    if market not in ["台股", "美股"]:
        raise ValueError("市場請填台股或美股")

    symbol = symbol.strip().upper()
    name = name.strip()

    if not symbol or not name:
        raise ValueError("股票代號與名稱不可空白")

    try:
        conn.execute("""
            INSERT INTO stocks
            (market, symbol, name, sector, created_at)
            VALUES (?, ?, ?, ?, ?)
        """, (
            market, symbol, name, sector,
            datetime.now().isoformat(timespec="seconds")
        ))
        conn.commit()
        print(f"新增成功：{symbol} {name}")

    except sqlite3.IntegrityError:
        print("新增失敗：這個市場的股票代號已存在")


# =========================
# 3. 查看股票清單
# =========================

def show_stocks():
    df = pd.read_sql_query("""
        SELECT id, market, symbol, name, sector
        FROM stocks
        ORDER BY market, symbol
    """, conn)

    display(df)
    return df


# =========================
# 4. 刪除股票
# =========================

def delete_stock(stock_id):
    cursor = conn.execute(
        "DELETE FROM stocks WHERE id = ?",
        (int(stock_id),)
    )
    conn.commit()

    if cursor.rowcount:
        print("股票已刪除")
    else:
        print("找不到這個股票 ID")


# =========================
# 5. 新增交易紀錄
# =========================

def add_trade(stock_id, trade_date, side,
              quantity, price, fee=0, note=""):

    if side not in ["買進", "賣出"]:
        raise ValueError("交易類型請填買進或賣出")

    datetime.strptime(trade_date, "%Y-%m-%d")

    quantity = float(quantity)
    price = float(price)
    fee = float(fee)

    if quantity <= 0 or price < 0 or fee < 0:
        raise ValueError("數量必須大於 0，價格與手續費不可為負")

    conn.execute("""
        INSERT INTO trades
        (stock_id, trade_date, side, quantity, price, fee, note)
        VALUES (?, ?, ?, ?, ?, ?, ?)
    """, (
        int(stock_id), trade_date, side,
        quantity, price, fee, note
    ))

    conn.commit()
    print("交易紀錄新增成功")


# =========================
# 6. 查看交易紀錄
# =========================

def show_trades():
    df = pd.read_sql_query("""
        SELECT
            t.id,
            s.market,
            s.symbol,
            s.name,
            t.trade_date,
            t.side,
            t.quantity,
            t.price,
            t.fee,
            t.quantity * t.price AS amount,
            t.note
        FROM trades t
        JOIN stocks s ON t.stock_id = s.id
        ORDER BY t.trade_date DESC, t.id DESC
    """, conn)

    display(df)
    return df


# =========================
# 7. 新增每日筆記
# =========================

def add_note(note_date, title, content, stock_id=None):
    datetime.strptime(note_date, "%Y-%m-%d")

    if not title.strip() or not content.strip():
        raise ValueError("筆記標題與內容不可空白")

    conn.execute("""
        INSERT INTO notes
        (stock_id, note_date, title, content)
        VALUES (?, ?, ?, ?)
    """, (
        int(stock_id) if stock_id not in (None, "") else None,
        note_date, title.strip(), content.strip()
    ))

    conn.commit()
    print("筆記新增成功")


# =========================
# 8. 查看每日筆記
# =========================

def show_notes():
    df = pd.read_sql_query("""
        SELECT
            n.id,
            n.note_date,
            s.symbol,
            s.name,
            n.title,
            n.content
        FROM notes n
        LEFT JOIN stocks s ON n.stock_id = s.id
        ORDER BY n.note_date DESC, n.id DESC
    """, conn)

    display(df)
    return df


# =========================
# 9. 匯出 CSV，供 Excel 使用
# =========================

def export_csv():
    queries = {
        "stocks.csv": "SELECT * FROM stocks",
        "trades.csv": """
            SELECT t.*, s.symbol, s.name
            FROM trades t
            JOIN stocks s ON t.stock_id = s.id
        """,
        "notes.csv": """
            SELECT n.*, s.symbol, s.name
            FROM notes n
            LEFT JOIN stocks s ON n.stock_id = s.id
        """
    }

    for filename, query in queries.items():
        df = pd.read_sql_query(query, conn)
        df.to_csv(
            "/content/" + filename,
            index=False,
            encoding="utf-8-sig"
        )
        print("已匯出：", filename)

    print("請到 Colab 左側「檔案」面板下載 CSV。")


print("所有功能已載入，可以開始使用！")
```

```python

# 新增兩支股票
add_stock("台股", "2330", "台積電", "半導體")
add_stock("美股", "AAPL", "Apple", "科技")

# 查看股票清單
show_stocks()

# 新增交易紀錄
# 先確認 show_stocks() 顯示的股票 ID
add_trade(
    stock_id=1,
    trade_date="2026-10-26",
    side="買進",
    quantity=10,
    price=1500,
    fee=5,
    note="測試交易"
)

# 查看交易紀錄
show_trades()

# 新增每日筆記
add_note(
    note_date="2026-10-26",
    title="今日市場觀察",
    content="記錄今天的市場變化與投資想法。",
    stock_id=1
)

# 查看筆記
show_notes()

# 匯出 Excel 可讀取的 CSV
export_csv()
```
